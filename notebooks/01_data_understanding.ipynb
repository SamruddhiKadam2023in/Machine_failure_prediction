## 2.5 How the raw data represents each failure type
There is no single failure-type column. No failure = all flags 0; each mode is its own 0/1 column.

```python
n_modes = df[FAILURE_MODE_COLUMNS].sum(axis=1)
pd.crosstab(df[MACHINE_FAILURE_COLUMN], n_modes, rownames=['Machine failure'], colnames=['mode flags set'])
```

```python
def combo(row):
    active = [m for m in FAILURE_MODE_COLUMNS if row[m] == 1]
    return '+'.join(active) if active else 'none'

pd.DataFrame({'Machine failure': df[MACHINE_FAILURE_COLUMN], 'modes': df[FAILURE_MODE_COLUMNS].apply(combo, axis=1)}).value_counts().rename('count').reset_index()
```

## Class distribution under the approved target definition
Preview only — the mapping is implemented in `src/preprocessing.py` in Stage 3. RNF is excluded as a class; rows with contradictory, unknown or multiple mode labels are excluded.

```python
modes4 = ['TWF', 'HDF', 'PWF', 'OSF']
n4 = df[modes4].sum(axis=1)
excluded = ((df[MACHINE_FAILURE_COLUMN] == 0) & (df['RNF'] == 1)) | ((df[MACHINE_FAILURE_COLUMN] == 1) & (n4 == 0)) | (n4 > 1)
kept = df[~excluded]
target = pd.Series('No Failure', index=kept.index)
for m in modes4:
    target[kept[m] == 1] = m
dist = target.value_counts().rename('count').to_frame()
dist['percent'] = (100 * dist['count'] / len(target)).round(2)
print('Excluded rows:', int(excluded.sum()), '| Kept rows:', len(kept))
dist
```
